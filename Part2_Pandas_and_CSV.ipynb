**Example 18 — Full Mini Pipeline**

Concept End-to-End Data Flow

```python
import pandas as pd

# read source data
df = pd.read_csv("orders.csv")

# clean province
df["province"] = (
    df["province"].str.upper()
)

# fill null status
df["status"] = df["status"].fillna(
    "UNKNOWN"
)

# calculate VAT
df["vat"] = df["amount"] * 0.07

# export output
df.to_csv(
    "processed_orders.csv",
    index=False
)

print(df)
print("Pipeline Completed")
```

**Example 7 — Function**
---
*   Concept สร้าง reusable logic
*   Function = reusable transformation logic

```python
def calculate_vat(amount):
    vat = amount * 0.07
    return vat
#===========
# main
#===========
a = 500.0
result = calculate_vat(a)
print(result)
```

**Example 8 — Mini Order Pipeline**
---
*   Concept เห็น Data Flow

```python
# source data

order = {
    "order_id": "ORD001",
    "amount": 1500
}

# transformation

vat = order["amount"] * 0.07
net_amount = order["amount"] + vat

# output

print("Order ID:", order["order_id"])
print("Net Amount:", net_amount)
```
